# 1. Was ist ein Notebook?

Ein Notebook verbindet **Text, Formeln, Programmcode und Ergebnisse** in einem Dokument.

Ein typischer technischer Ablauf lautet:

**Aufgabe → Gegeben → Gesucht → Formel → Python-Berechnung → Ergebnis → Plausibilitätsprüfung → Diagramm**

### Jupyter und Google Colab

- **Jupyter Notebook** ist eine Umgebung für interaktive Dokumente mit ausführbarem Code.
- **Google Colab** ist eine browserbasierte Jupyter-Umgebung.
- Das Notebook-Dateiformat ist meist `.ipynb`.
- Eine **Zelle** ist ein einzelner Baustein des Notebooks.
- Ein **Kernel / eine Laufzeit** führt Python aus.

### Zwei zentrale Zelltypen

| Zelltyp | Zweck |
|---|---|
| Markdown/Text | Erklärungen, Überschriften, Tabellen, Formeln |
| Code | Python-Anweisungen und Berechnungen |

**Wichtig:** Eine Code-Zelle kann nur etwas berechnen, wenn die benötigten Variablen und Bibliotheken in der aktuellen Laufzeit vorhanden sind.

````python
print("Hallo! Das ist meine erste ausgeführte Python-Zelle.")
````

# 2. Die wichtigsten Notebook-Handgriffe

- **Zelle auswählen:** anklicken.
- **Zelle ausführen:** `Shift + Enter`.
- **Neue Code-Zelle:** in Colab über `+ Code`.
- **Neue Textzelle:** in Colab über `+ Text`.
- **Zelltyp wechseln:** über die Zell-/Menüoptionen.
- **Laufzeit neu starten:** verwenden, wenn der Zustand unübersichtlich geworden ist.

### Ausführungsreihenfolge

Zellen werden **nicht automatisch in der sichtbaren Reihenfolge ausgeführt**. Wenn Zelle 3 eine Variable aus Zelle 1 benötigt, muss Zelle 1 vorher ausgeführt worden sein.

> **Gute Gewohnheit:** Vor der Abgabe eines Notebooks die Laufzeit neu starten und alle Zellen von oben nach unten ausführen.

## 3.1 Markdown und technische Formeln

Mathematische Formeln werden in Markdown mit LaTeX geschrieben.

Inline:

```text
$I = \frac{U}{R}$
```

Eigene Formelzeile:

```text
$$
U = R \cdot I
$$
```

Häufige Befehle:

| LaTeX | Bedeutung |
|---|---|
| `x^2` | Hochzahl |
| `x_1` | Index |
| `\frac{a}{b}` | Bruch |
| `\sqrt{x}` | Wurzel |
| `\cdot` | Multiplikationspunkt |
| `\Delta` | Delta |
| `\le` | kleiner oder gleich |
| `\ge` | größer oder gleich |

Für Temperaturangaben ist `°` in Colab robust, z. B. `$20\,°\mathrm{C}$` oder einfach `20 °C`.

# 6. Ausgaben formatieren

Mit f-Strings lassen sich Werte lesbar ausgeben:

```python
f"U = {U:.2f} V"
```

`.2f` bedeutet: zwei Nachkommastellen.

````python
U = 12.0
R = 6.0
I = U / R
print(f"U = {U:.2f} V")
print(f"R = {R:.2f} Ohm")
print(f"I = {I:.2f} A")
````

# 7. Technischer Rechenweg: Ohmsches Gesetz

### Gegeben

$U = 12\,V$

$R = 6\,Ω$

### Gesucht

$I$

### Formel

$$
I = \frac{U}{R}
$$

### Python

````python
U = 12.0       # V
R = 6.0        # Ohm
I = U / R
print(f"I = {I:.2f} A")
````

> **Merke:** Python kennt die physikalische Einheit nicht automatisch. Dokumentiere Einheiten im Kommentar, Variablennamen oder im erklärenden Text.

# 8. Vergleiche und Bedingungen

Vergleichsoperatoren:

`==` gleich · `!=` ungleich · `<` kleiner · `>` größer · `<=` kleiner/gleich · `>=` größer/gleich

Eine Bedingung liefert `True` oder `False`.

````python
spannung = 5.0
print(spannung > 3)
print(spannung == 5)
print(spannung < 2)
````

````python
temperatur = 25

if temperatur < 10:
    print("niedrig")
elif temperatur < 30:
    print("mittel")
else:
    print("hoch")
````

# 9. Listen

Listen speichern mehrere Werte:

```python
spannungen = [1.0, 2.0, 3.0, 4.0]
```

Der erste Index ist `0`.

````python
spannungen = [1.0, 2.0, 3.0, 4.0]
print(spannungen)
print("erster:", spannungen[0])
print("letzter:", spannungen[-1])
print("Anzahl:", len(spannungen))
spannungen.append(5.0)
print(spannungen)
````

# 10. Schleifen

Mit `for` werden Werte nacheinander verarbeitet.

````python
spannungen = [1.0, 2.0, 3.0, 4.0]
for U in spannungen:
    print(f"U = {U:.1f} V")
````

`range(5)` erzeugt die Werte `0, 1, 2, 3, 4`.

````python
for n in range(5):
    print(n)
````

# 12. Funktionen

Funktionen bündeln wiederverwendbare Arbeitsschritte.

```python

```

````python
def ohmsches_gesetz(U, R):
    return U / R

I = ohmsches_gesetz(12.0, 6.0)
print(f"I = {I:.2f} A")
````

# 13. Fehler verstehen und Debugging

Fehler gehören zum Programmieren.

| Fehlertyp | Typische Ursache |
|---|---|
| `SyntaxError` | Schreibweise ungültig |
| `NameError` | Variable/Name unbekannt |
| `TypeError` | unpassende Datentypen |
| `ZeroDivisionError` | Division durch 0 |
| `IndexError` | Listenindex existiert nicht |

### Systematische Fehlersuche

1. Lies die letzte Zeile der Fehlermeldung.
2. Prüfe Eingabewerte mit `print()`.
3. Prüfe Datentypen mit `type()`.
4. Prüfe Zwischenergebnisse.
5. Vergleiche Python-Ausdruck und mathematische Formel.
6. Prüfe Einheiten und Vorzeichen.
7. Prüfe die Ausführungsreihenfolge der Zellen.

````python
# Absichtlicher Fehler: Was passiert hier?
zahl = 10
print(zahl / 0)
````

Behebe den Fehler selbst und führe die Zelle erneut aus.

> **Wichtig:** Nicht einfach alles neu schreiben. Fehlermeldungen sind Hinweise darauf, wo du suchen solltest.

# 16. Matplotlib und Diagramme

Ein technisches Diagramm sollte mindestens sinnvolle Achsenbeschriftungen und, wenn nötig, Einheiten enthalten.

````python
import numpy as np
import matplotlib.pyplot as plt

R = 100.0  # Ohm
I = np.linspace(0, 0.1, 100)
U = R * I

plt.plot(I, U)
plt.xlabel("Strom I [A]")
plt.ylabel("Spannung U [V]")
plt.title("Ohmsche Kennlinie")
plt.grid(True)
plt.show()
````

# 17. Runden und Genauigkeit

`round(wert, 2)` rundet einen Wert auf zwei Nachkommastellen.

Bei der Ausgabe kann man ebenfalls formatieren:

```python
f"{wert:.2f}"
```

> **Merke:** Darstellung und Berechnung sind nicht dasselbe. Wenn eine Aufgabe erst am Ende runden verlangt, während der Berechnung nicht vorzeitig runden.

````python
wert = 3.14159265
print(wert)
print(round(wert, 2))
print(f"{wert:.4f}")
````

# 21. Plausibilitätsprüfung

Python kann rechnen, aber nicht automatisch entscheiden, ob eine technische Lösung sinnvoll ist.

Prüfe:

- Sind die Einheiten korrekt?
- Stimmt die Größenordnung?
- Ist das Vorzeichen plausibel?
- Wurde die richtige Formel verwendet?
- Wurden mV und V bzw. mA und A verwechselt?
- Wurde zu früh gerundet?
- Passt das Ergebnis zur Aufgabenstellung?

**Python ist ein Rechenwerkzeug, kein Ersatz für technische Plausibilitätsprüfung.**

# 23. Vollständiges Mini-Beispiel

## Aufgabe

Ein Widerstand von $R = 220\,Ω$ wird mit $I = 20\,mA$ betrieben. Gesucht ist $U$.

## Formel

$$
U = R \cdot I
$$

````python
# Gegebene Werte
R = 220.0       # Ohm
I = 20e-3       # A

# Berechnung
U = R * I

# Ergebnis
print(f"R = {R:.1f} Ohm")
print(f"I = {I:.3f} A")
print(f"U = {U:.2f} V")
````

## Interpretation

Das Ergebnis liegt im Bereich weniger Volt und ist für `220 Ω` bei `20 mA` plausibel.

# 28. Häufige Fragen

### Warum wird mein Code nicht ausgeführt?
Prüfe Zelltyp, Laufzeit, Ausführung und Fehlermeldung.

### Warum ist eine Variable unbekannt?
Die Definition wurde möglicherweise nicht ausgeführt oder die Laufzeit wurde neu gestartet.

### Warum funktioniert `3,14` nicht als Dezimalzahl?
Python verwendet `3.14`.

### Warum muss ich bei `if` einrücken?
Die Einrückung gehört zur Python-Syntax.

### Warum beginnt eine Liste bei 0?

### Warum ist ein Diagramm leer?
